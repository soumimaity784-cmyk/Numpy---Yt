```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
print("Shape:", data.shape)
```

```python
#Find the average age of students.
print("Average age:", np.mean(data[:, 0]))
```

```python
#Extract Math marks of all students.
print("Math marks:", data.T[1])
```

```python
#Find the highest Science mark.
print("Highest Science mark:", np.amax(data[:, 2]))
```

```python
#Get details of the student who scored more than 90 in Math
idx = np.where(data[:, 1] > 90)[0]
print("Student number:", idx + 1)
print("[Age, Math, Science]:", data[idx])
```

```python
#Increase Math marks of all students by 5.
bumped = data.copy()
bumped[:, 1] = bumped[:, 1] + 5
print("New Math marks:", bumped[:, 1])
```

```python
#Find how many students are younger than 19.
print("Students younger than 19:", np.count_nonzero(data[:, 0] < 19))
```

```python
#Calculate the average marks in each subject (column-wise mean)
np.mean(data[:, 1:], axis=0)
```

```python
#Get data of students who scored at least 80 in both subjects.
data[(data[:, 1] >= 80) & (data[:, 2] >= 80)]
```

```python
#Replace all Science marks < 75 with 0.
data[:, 2] = np.where(data[:, 2] < 70, 0, data[:, 2])
data
```
